```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
movies = pd.read_csv("C:/Users/HP/Downloads/archive (3)/Zomato Restaurant Dataset.csv")
```

```python
movies.info
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(5,5))
plt.hist(
    movies["Latitude"],
    bins=8,
    color="orange",
    edgecolor="black",
)
plt.title("Distribution of Latitude")
plt.xlabel("runtime min")
plt.ylabel("Number of Latitude")
plt.show()
```

```python
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))
sns.histplot(
    movies["Longitude"],
    bins=5,
    color="lightgreen",
    edgecolor="black",
)
plt.title("Distribution of Longtitude")
plt.xlabel("runtime min")
plt.ylabel("Number of Longtitude")
plt.show()
```

```python
plt.figure(figsize=(5,5))
sns.countplot(data=movies, x="Restaurant ID")
plt.title("Number of Restaurant ID")
plt.xlabel("Genre")
plt.ylabel("Number of ID")
plt.xticks(rotation=32)
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.scatter(movies["Restaurant ID"],movies["Latitude"],color="green",alpha=0.7)

plt.title("Budget vs Box-office Collection")
plt.xlabel("Budget($ crore)")
plt.ylabel("Worldwide gross($ crore)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.regplot(
    data=movies, x="Latitude",y="Aggregate rating",
    scatter_kws={"alpha": 0.6 },line_kws={"color":"violet"},)
plt.title("Budget vs Collection with Trend Line")
plt.show()
```

```python
plt.figure(figsize=(5,6))
sns.boxplot(
    data=movies, x="Latitude",y="Votes",
    color="blue",
)
plt.title("Box-office collection by industry")
plt.ylabel("Worldwide gross($crore)")
plt.show()
```

```python
plt.figure(figsize=(9,9))
plt.scatter(movies["Votes"],movies["City"],color="green",alpha=0.9)

plt.title("Budget vs Box-office Collection")
plt.xlabel("Budget($ crore)")
plt.ylabel("Worldwide gross($ crore)")
plt.show()
```
